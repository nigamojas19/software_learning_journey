```python
class Student:
    #Attributes
    name=""
    age=0
    gender=""
    #methods
    def set_info(self):
        self.name=input("Enter your name = ")
        self.age=int(input("Enter your age = "))
        self.gender=input("Enter your Gender = ")
    def display(self):
        print(f"My name is {self.name}, age is {self.age} and gender is {self.gender}")
```

```python
s1=Student()
s1.set_info()
```

```python
s1.display()
```
