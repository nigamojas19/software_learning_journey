```python
class Student:          
    #methods
    def __init__(self, name: str, age = int, gender = str) -> None:
        print("Constructor")
        self.name=name
        self.age=age
        self.gender=gender
    def display(self) -> None:
        print(f"My name is {self.name}, age is {self.age} and gender is {self.gender}")
```

```python
s1=Student("ojas", 21, "male")
```

```python
s1.display()
```
